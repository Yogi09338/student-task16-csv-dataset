```python
# Day 16 - Load a CSV Dataset
# AI & ML Internship

import pandas as pd

# Step 1: Create a sample employee dataset
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Arun", "Bharath", "Chetan", "Divya", "Kiran",
        "Manoj", "Neha", "Pooja", "Rahul", "Sneha"
    ],
    "Department": [
        "AIML", "CSE", "ISE", "AIML", "CSE",
        "ISE", "AIML", "CSE", "ISE", "AIML"
    ],
    "Salary": [
        35000, 42000, 38000, 45000, 40000,
        36000, 48000, 39000, 43000, 50000
    ]
}

# Step 2: Convert the data into a DataFrame
df = pd.DataFrame(data)

# Step 3: Save the DataFrame as a CSV file
csv_file = "employee_data.csv"
df.to_csv(csv_file, index=False)

print("CSV dataset created successfully!")
print("File name:", csv_file)

# Step 4: Load the CSV dataset using Pandas
df = pd.read_csv(csv_file)

print("\nCSV dataset loaded successfully!")

# Step 5: Display the complete dataset
print("\nComplete Dataset:")
print(df)

# Step 6: Display the first 5 records
print("\nFirst 5 Records:")
print(df.head())

# Step 7: Display the last 5 records
print("\nLast 5 Records:")
print(df.tail())
```
